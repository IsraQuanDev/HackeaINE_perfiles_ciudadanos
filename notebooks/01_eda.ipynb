# Análisis Exploratorio de Datos (EDA): Participación Electoral Municipal 2009–2024

**Proyecto:** Perfiles Ciudadanos de Participación Electoral (Hackatón INE - Reto 1)  
**Nota metodológica:** Los indicadores y perfiles aquí analizados describen **territorios (municipios)**, nunca individuos, evitando la falacia ecológica.

```python
import json
from pathlib import Path
import pandas as pd
import numpy as np

FEAT_PATH = Path('../data/processed/features_municipio.parquet')
df = pd.read_parquet(FEAT_PATH)
print(f'Municipios analizados: {len(df):,} | Variables: {df.shape[1]}')
df.head()
```

## 1. Distribuciones de Participación por Año y Tipo de Elección (2009–2024)

```python
cols_part = ['part_2009', 'part_2012', 'part_2015', 'part_2018', 'part_2021', 'part_2024',
             'part_media_presidencial', 'part_media_intermedia', 'caida_presidencial_intermedia']
df[cols_part].describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9]).T
```

## 2. Matriz de Correlaciones entre Dinámica Electoral, Brecha de Género, Estructura Etaria y Urbanización

```python
corr_cols = [
    'part_media_global', 'caida_presidencial_intermedia', 'pendiente_tendencia',
    'volatilidad', 'constancia_sobre_mediana', 'brecha_genero_media',
    'brecha_generacional_60_vs_18_29', 'pct_ln_jovenes_18_29', 'pct_ln_adultos_60_mas',
    'pct_ln_urbana', 'pct_ln_rural', 'tasa_voto_anticipado_por_10k'
]
df[corr_cols].corr().round(3)
```

## 3. Autocorrelación Espacial (Moran's I Global y Clusters LISA Locales)

```python
print('Moran I Global:', df['moran_global_i'].iloc[0], '| p-valor:', df['moran_global_p_val'].iloc[0])
df.groupby('lisa_cluster')[['part_media_global', 'caida_presidencial_intermedia', 'pct_ln_rural']].agg(['count', 'mean'])
```
